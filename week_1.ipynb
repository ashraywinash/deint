```python
# WEEK-1: Basic Data Handling Commands

import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['target'] = iris.target

```

```python
# 2. Dimension of the data
print("Shape of the data (rows, columns):", df.shape)

```

```python
# 3. Display data (top 5 rows and total data)
print("\nTop 5 rows:\n", df.head())
print("\nTotal data:\n", df)
```

```python

# 4. List the column names of a data frame
print("\nColumn Names:\n", df.columns.tolist())
```

```python

# 5. Change columns of a data frame (example: renaming a column if needed)
df.rename(columns={df.columns[0]: "new_column_name"}, inplace=True)

```

```python
# 6. Display specific single column or multiple columns of a data frame
print("\nSingle Column:\n", df[df.columns[0]])  
print("\nMultiple Columns:\n", df[df.columns[0:2]])

```

```python
# 7. Bind sets of rows of dataframes (row-wise concat)
df_extra = df.copy()  
df_rows_combined = pd.concat([df, df_extra], axis=0)
print("\nRow-bound Data Shape:", df_rows_combined.shape)
```

```python
# 8. Bind sets of columns of data frames (column-wise concat)
df_new_cols = df.copy()
df_cols_combined = pd.concat([df, df_new_cols], axis=1)
print("\nColumn-bound Data Shape:\n", df_cols_combined.shape)

```

```python
# 9. Find missing values in the dataset
print("\nMissing Values in Dataset:\n", df.isnull().sum())
```
