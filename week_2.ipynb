```python
import seaborn as sns
import pandas as pd
from scipy import stats

# Load the Iris dataset
df = sns.load_dataset("iris")
print(df.head())
```

```python
# Mean
print("Mean of features:")
print(df.mean(numeric_only=True))

# Median
print("\nMedian of features:")
print(df.median(numeric_only=True))

# Mode
print("\nMode of features:")
print(df.mode(numeric_only=True).iloc[0])  # only the first mode
```

```python
print("\nRange of each feature (max - min):")
range_vals = df.max(numeric_only=True) - df.min(numeric_only=True)
print(range_vals)
```

```python
# Variance
print("\nVariance of each feature:")
print(df.var(numeric_only=True))

# Standard Deviation
print("\nStandard Deviation of each feature:")
print(df.std(numeric_only=True))
```

```python
# Quartiles
print("\nQuartiles (25%, 50%, 75%):")
print(df.describe(percentiles=[.25, .5, .75]))

# Interquartile Range (IQR)
print("\nInterquartile Range (IQR):")
Q1 = df.quantile(0.25, numeric_only=True)
Q3 = df.quantile(0.75, numeric_only=True)
IQR = Q3 - Q1
print(IQR)
```
