```python
import pandas as pd
import seaborn as sns 

# Load dataset
df = sns.load_dataset('titanic')
df.head()
```

```python
# Dummy coding 'embarked' (nominal variable)
df_dummy = pd.get_dummies(df, columns=['embarked'], drop_first=True)

print(df_dummy[['embarked_Q', 'embarked_S']].head())
```

```python
import numpy as np
from sklearn.preprocessing import OrdinalEncoder

# Sample data
data = pd.DataFrame({'education': ['High School', 'Bachelors', 'Masters', 'PhD', 'Bachelors']})

# Define order
encoder = OrdinalEncoder(categories=[['High School', 'Bachelors', 'Masters', 'PhD']])
data['education_encoded'] = encoder.fit_transform(data[['education']])

print(data)
```

```python
from sklearn.datasets import load_iris

# Load dataset
iris = load_iris(as_frame=True)
df = iris.frame

# Convert sepal length into categories (equal-width binning)
df['sepal_length_cat'] = pd.cut(df['sepal length (cm)'], bins=3, labels=['Short', 'Medium', 'Long'])

print(df[['sepal length (cm)', 'sepal_length_cat']].head())
```
