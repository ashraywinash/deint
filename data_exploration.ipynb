```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Iris dataset
iris = sns.load_dataset('iris')
print(iris.head())
```

```python
# List of all predictor columns
predictors = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

# Create individual boxplots
for col in predictors:
    sns.boxplot(x=iris[col])
    plt.title(f'Box plot of {col}')
    plt.show()
```

```python
sns.boxplot(x=iris['sepal_width'])
plt.title('Box plot of Sepal Width')
plt.show()
```

```python

sns.histplot(iris['petal_length'], kde=True)
plt.title('Histogram of Petal Length')
plt.xlabel('Petal Length (cm)')
plt.show()
```

```python
sns.scatterplot(x='sepal_length', y='petal_length', hue='species', data=iris)
plt.title('Petal Length vs Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.show()
```
