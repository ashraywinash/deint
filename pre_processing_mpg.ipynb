```python
import pandas as pd
import seaborn as sns
import numpy as np

# Load Auto MPG dataset
df = sns.load_dataset('mpg')

# Display the first few rows
print(df.head())
```

```python
# Step 1: Check for missing values
print(df.isnull().sum())

# Remove rows with missing values
df_cleaned = df.dropna()

# Optional: Remove outliers in 'mpg' using IQR
Q1 = df_cleaned['mpg'].quantile(0.25)
Q3 = df_cleaned['mpg'].quantile(0.75)
IQR = Q3 - Q1

# Define outlier bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter out outliers
df_no_outliers = df_cleaned[(df_cleaned['mpg'] >= lower_bound) & (df_cleaned['mpg'] <= upper_bound)]

print(f"Original rows: {len(df)}, After cleaning: {len(df_no_outliers)}")
```

```python
# Introduce a missing value artificially for demonstration
df_impute = df.copy()
df_impute.loc[10:12, 'horsepower'] = np.nan  # simulate missing values

# Impute missing 'horsepower' values with mean
mean_hp = df_impute['horsepower'].mean()
df_impute['horsepower'] = df_impute['horsepower'].fillna(mean_hp)

print(df_impute.loc[10:12, ['horsepower']])
```

```python
# Cap 'weight' at 95th percentile
cap_value = df_no_outliers['weight'].quantile(0.95)

# Replace values above the cap with the cap
df_no_outliers['weight_capped'] = np.where(df_no_outliers['weight'] > cap_value, cap_value, df_no_outliers['weight'])

print(df_no_outliers[['weight', 'weight_capped']].head())
```
